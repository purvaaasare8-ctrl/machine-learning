```
import pandas as pd

df = pd.read_csv("sample.csv", header=0)
```

```
X = df.drop("Cost Price", axis=1)

y = df["Cost Price"]

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

```
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("/content/sample.csv").iloc[1:]

numerical_cols = ['Opening', 'Purchase/', 'Number of', 'Hand-In-', 'Cost Price.1']
df[numerical_cols] = df[numerical_cols].apply(pd.to_numeric)

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df[numerical_cols])

scaled_df = pd.DataFrame(scaled_data, columns=numerical_cols)

print("Scaled Data Head:")
display(scaled_df.head())
```

```
categorical_cols = X.select_dtypes(include="object").columns

X_encoded = pd.get_dummies(X, columns=categorical_cols)
X_encoded = X_encoded.astype(int)

print(X_encoded.head())
```

```
from sklearn.preprocessing import StandardScaler
import pandas as pd

cols = ['Opening', 'Purchase/', 'Number of', 'Hand-In-']

data = X.iloc[1:].copy()
data[cols] = data[cols].astype(float)

scaler = StandardScaler()
data[cols] = scaler.fit_transform(data[cols])

print("Data scaled successfully:")
display(data[cols].head())
```

```
from sklearn.model_selection import train_test_split

y_clean = y.iloc[1:]

X_train, X_test, y_train, y_test = train_test_split(
    data,
    y_clean,
    test_size=0.33,
    random_state=42
)

print(f"X shape: {data.shape}")
print(f"y shape: {y_clean.shape}")
print("Split successful!")
```

```
from sklearn.linear_model import LinearRegression

X_train_numeric = X_train.drop(columns=['Product ID', 'Product Name'])

model = LinearRegression()

model.fit(X_train_numeric, y_train.astype(float))

print("Model trained successfully!")
print("Intercept:", model.intercept_)
print("Coefficients:")
for col, coef in zip(X_train_numeric.columns, model.coef_):
    print(f"{col}: {coef}")
```

```
X_test_numeric = X_test.drop(columns=['Product ID', 'Product Name'])

y_pred = model.predict(X_test_numeric)

print("Predicted Values:")
print(y_pred[:10])

print("\nActual Values:")
print(y_test.iloc[:10].values.astype(float))
```

```
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)

if r2 >= 0.8:
    print("The model performs very well.")
elif r2 >= 0.6:
    print("The model performs well.")
else:
    print("The model needs improvement.")
```

```
import joblib

joblib.dump(model, "LR_model.pkl")
joblib.dump(scaler, "scaler.pkl")
joblib.dump(X_encoded.columns.tolist(), "columns.pkl")

print("Files saved successfully.")
```
